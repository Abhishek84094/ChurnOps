## 4. Target Distribution — Class Imbalance

```python
target_col = PARAMS['dataset']['target_column']
counts = df[target_col].value_counts()
pcts   = df[target_col].value_counts(normalize=True) * 100

print('Target distribution:')
for cls in counts.index:
    print(f'  {cls}: {counts[cls]:>5}  ({pcts[cls]:.2f}%)')
print(f'\nImbalance ratio (No:Yes): {counts["No"]/counts["Yes"]:.2f}:1')
print('\nWhy accuracy is misleading:')
print(f'  A dummy classifier predicting always-No scores {pcts["No"]:.1f}% accuracy.')
print('  We must use Recall, F1, and PR-AUC as primary metrics.')

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Bar chart
axes[0].bar(counts.index, counts.values, color=['#2196F3', '#F44336'], alpha=0.85, edgecolor='white')
for i, (cls, cnt) in enumerate(counts.items()):
    axes[0].text(i, cnt + 30, f'{cnt}\n({pcts[cls]:.1f}%)', ha='center', fontsize=11, fontweight='bold')
axes[0].set_title('Churn Class Counts', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Count')
axes[0].set_ylim(0, counts.max() * 1.2)
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)

# Pie chart
axes[1].pie(counts.values, labels=counts.index, autopct='%1.1f%%',
            colors=['#2196F3', '#F44336'], startangle=90,
            wedgeprops=dict(edgecolor='white', linewidth=2))
axes[1].set_title('Churn Distribution', fontsize=13, fontweight='bold')

plt.tight_layout()
plt.savefig(PROJECT_ROOT / 'reports' / 'eda' / 'target_distribution.png', bbox_inches='tight')
plt.show()
print('Saved -> reports/eda/target_distribution.png')
```

## 5. TotalCharges — Blank String Investigation
Validation detected 11 blank strings. Let's inspect them.

```python
blank_mask = df['TotalCharges'].str.strip() == ''
print(f'Rows with blank TotalCharges: {blank_mask.sum()}')

cols_to_show = ['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']
blank_rows = df[blank_mask][cols_to_show]
print()
print(blank_rows.to_string(index=False))

print(f'\nConclusion: All {blank_mask.sum()} rows have tenure=0.')
print('These are brand-new customers who have not been billed yet.')
print('Strategy (Phase 4): fill TotalCharges = MonthlyCharges * tenure = 0 OR impute as 0.')
print('This is a DETERMINISTIC rule, not a statistical imputation -> no leakage risk.')
```

## 6. SeniorCitizen — Encoding Issue

```python
print('SeniorCitizen unique values:', df['SeniorCitizen'].unique())
print('All other binary columns use Yes/No, SeniorCitizen uses 0/1.')
print('Action (Phase 4): map 0->No, 1->Yes for consistency.')
print()
print(df['SeniorCitizen'].value_counts())
```
